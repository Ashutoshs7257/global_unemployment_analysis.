```python
# Top 10 countries by average unemployment
top10_avg_unemployment = (
    df_unemployment
    .groupby('Country')['Value']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(top10_avg_unemployment)
```

```python
# Bottom 10 countries by average unemployment.
bottom10_avg_unemployment = (
    df_unemployment
    .groupby('Country')['Value']
    .mean()
    .sort_values(ascending=True)
    .head(10)
)

print(bottom10_avg_unemployment)
```

```python
# Year with highest global unemployment
global_avg_yearly = (
    df_unemployment
    .groupby('year')['Value']
    .mean()
)

highest_year = global_avg_yearly.idxmax()

print("Year with Highest Global Unemployment:", highest_year)
print("Average Unemployment:", global_avg_yearly.max())
```

```python
# Year with highest global unemployment
global_avg_yearly = (
    df_unemployment
    .groupby('year')['Value']
    .mean()
)

lowest_year = global_avg_yearly.idxmin()

print("Year with Lowest Global Unemployment:", lowest_year)
print("Average Unemployment:", global_avg_yearly.min())
```

```python
# most stable country
stable_country = (
    df_unemployment
    .groupby('Country')['Value']
    .std()
    .sort_values()
    .head(1)
)

print("Most Stable Country (Lowest Variation):")
print(stable_country)
```

```python
# most 
volatile_country = (
    df_unemployment
    .groupby('Country')['Value']
    .std()
    .sort_values(ascending=False)
    .head(1)
)

print("Most Volatile Country (Highest Variation):")
print(volatile_country)
```

```python
# get average (safe in case multiple entries per country-year)
df_1991 = df_unemployment[df_unemployment['year'] == 1991].groupby('Country')['Value'].mean()
df_2024 = df_unemployment[df_unemployment['year'] == 2024].groupby('Country')['Value'].mean()

# combine into one DataFrame
improvement = pd.DataFrame({
    '1991': df_1991,
    '2024': df_2024
})

# drop missing values (countries not present in both years)
improvement = improvement.dropna()

# calculate improvement
improvement['Change'] = improvement['1991'] - improvement['2024']

# sort to find biggest improvement
top_improved = improvement.sort_values('Change', ascending=False)

print(top_improved.head(10))
```
